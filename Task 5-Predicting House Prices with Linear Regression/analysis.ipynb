# Handle missing values and encode categorical features (One-Hot Encoding)


```python
df['Date'] = pd.to_datetime(df['Date'])
df['Sale_Year'] = df['Date'].dt.year
df['Sale_Month'] = df['Date'].dt.month
df['Sale_Day'] = df['Date'].dt.day
df.drop('Date', axis=1, inplace=True)
```

# Train/test split (80/20)

```python
x = df.drop(columns=['Price', 'id','Sale_Year','Sale_Month','Sale_Day'])
y=df['Price']
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.20,random_state=43)
```

# Train a Linear Regression model using scikit-learn


```python
from sklearn.linear_model import LinearRegression
```

```python
model = LinearRegression()
```

```python
model.fit(x_train,y_train)
```

# Evaluate model using: Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R² score

```python
from sklearn.metrics import mean_squared_error,r2_score
```

```python
y_pred = model.predict(x_test)
```

```python
mse = mean_squared_error(y_test,y_pred)
mse
```

```python
rmse = np.sqrt(mse)
print(rmse)
```

```python
r2_sco = r2_score(y_test,y_pred)
r2_sco
```

# Scatter plot: actual prices vs. predicted prices (the closer to a diagonal line, the better)

```python
plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred,color='steelblue',alpha=0.6,edgecolor='black')
plt.plot([y_test.min(), y_test.max()],[y_test.min(), y_test.max()],color='red',linewidth=2,linestyle='--')

plt.title("Actual Prices vs Predicted Prices", fontsize=16)
plt.xlabel("Actual House Prices")
plt.ylabel("Predicted House Prices")

plt.grid(True)

plt.show()
```

# Residual plot: check that residuals are randomly distributed (no systematic pattern)


```python
residuals = y_test-y_pred
```

```python
plt.figure(figsize=(10,8))
plt.scatter(y_pred,residuals,color='steelblue',alpha=0.6,edgecolor='black')
plt.axhline(y=0,color='red',linestyle='--',linewidth=2)

plt.title("Residual Plot", fontsize=16)
plt.xlabel("Predicted House Prices")
plt.ylabel("Residuals (Actual - Predicted)")

plt.grid(True)

plt.show()
```

## Interpretation
##### The residuals are mostly scattered around the zero line, indicating that the Linear Regression model captures the overall relationship between the features and house prices reasonably well.
##### However, the spread of residuals increases for higher predicted house prices, suggesting the presence of heteroscedasticity (non-constant error variance).
##### A few large residuals (outliers) are also observed, indicating that the model predicts some high-value houses less accurately.

# Coefficient analysis: which features have the highest positive/negative impact on price?

```python
coef_df = pd.DataFrame({'Feature': x_train.columns,'Coefficient': model.coef_})
coef_df = coef_df.sort_values(by='Coefficient', ascending=False)
print("Feature Coefficients:")
display(coef_df)
```

```python
# Top 5 positive coefficients
top_positive = coef_df.head(5)

plt.figure(figsize=(10,6))
sns.barplot(data=top_positive,x='Coefficient',y='Feature')

plt.title("Top 5 Features with Positive Impact on House Price")
plt.xlabel("Coefficient")
plt.ylabel("Feature")
plt.show()

# Top 5 negative coefficients
top_negative = coef_df.tail(5)

plt.figure(figsize=(10,6))
sns.barplot(data=top_negative,x='Coefficient',y='Feature')

plt.title("Top 5 Features with Negative Impact on House Price")
plt.xlabel("Coefficient")
plt.ylabel("Feature")
plt.show()
```

### These are the 5 features having most postive and negative impact on House Price.

# Compare Linear Regression against a Ridge or Lasso regularised model


```python
from sklearn.linear_model import Ridge
```

```python
Ridge_model=Ridge(alpha=0.1)
```

```python
Ridge_model.fit(x_train,y_train)
```

```python
R_pred = Ridge_model.predict(x_test)
```

```python
comparison = pd.DataFrame({
    "Model": ["Linear Regression", "Ridge Regression"],
    "MSE": [
        mean_squared_error(y_test, y_pred),
        mean_squared_error(y_test, R_pred)
    ],
    "RMSE": [
        np.sqrt(mean_squared_error(y_test, y_pred)),
        np.sqrt(mean_squared_error(y_test, R_pred))
    ],
    "R² Score": [
        r2_score(y_test, y_pred),
        r2_score(y_test, R_pred)
    ]
})

comparison = comparison.round(4)

display(comparison)
```

## Comparison of Linear Regression and Ridge Regression

Both Linear Regression and Ridge Regression produced nearly identical performance metrics.

### Interpretation

- Both models explain approximately 70.53% of the variance in house prices.
- The RMSE values are almost identical, indicating very similar prediction accuracy.
- Ridge Regression did not improve the model because the dataset have limited multicollinearity and the Linear Regression model already generalizes well.

### Conclusion

The Linear Regression model performs as well as the Ridge Regression model for this dataset. The standard Linear Regression model is sufficient for predicting house prices in this analysis.
